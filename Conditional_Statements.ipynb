#**if-elif-else ladder**

create a ranking system:
 A(90-100),B(70-89),C(60-69),
 D(50-59),E(30-49),F(10-29)

```python
score=37
```

```python
if score>=90:
  print("Grade A")
elif score>=70 and score<=89:
  print("Grade B")
elif score>=60 and score<=69:
  print("Grade C")
elif score>=50 and score<=59:
  print("Grade D")
elif score>=30 and score<=49:
  print("Grade E")
else:
  print("Grade F")
```
